```python
vectors = []

for doc in docs:
    emb = client_openai.embeddings.create(
        model="text-embedding-3-large",
        input=doc["text"]
    ).data[0].embedding

    vectors.append({
        "table": doc["table"],
        "chunk_type": doc["chunk_type"],
        "text": doc["text"],
        "vector": emb
    })
```

```python
from qdrant_client.models import PointStruct
import uuid

points = []

for v in vectors:

    points.append(
        PointStruct(
            id=str(uuid.uuid4()),
            vector=v["vector"],
            payload={
                "table": v["table"],
                "chunk_type": v["chunk_type"],
                "content": v["text"]
            }
        )
    )

client.upsert(
    collection_name="schema_chunk_docs",
    points=points
)
```

```python
question = "مشتریان برتر از بابت بیشترین خرید در شعبه 2"
```

```python
client_openai = OpenAI(
    base_url=BASE_URL,
    api_key=API_KEY
)
question_embedding = client_openai.embeddings.create(
    model="text-embedding-3-large",
    input=question
).data[0].embedding
```

```python
results = client.query_points(
    collection_name="schema_chunk_docs",
    query=question_embedding,
    limit=15
)

for p in results.points:
    print(f"{p.payload['table']} ({round(p.score, 2)})")
```

```python
from collections import defaultdict

table_scores = defaultdict(float)

for p in results.points:
    table = p.payload["table"]

    table_scores[table] = max(
        table_scores[table],
        p.score
    )
```

```python
for table, score in sorted(
    table_scores.items(),
    key=lambda x: x[1],
    reverse=True
):
    print(table, round(score, 3))
```

```python
pairs = [
    [question, p.payload["content"]]
    for p in results.points
]
```

```python
from sentence_transformers import CrossEncoder

reranker = CrossEncoder(
    "BAAI/bge-reranker-v2-m3"
)
```

```python
rerank_scores = reranker.predict(pairs)
```

```python
reranked = []

for point, score in zip(results.points, rerank_scores):
    reranked.append({
        "table": point.payload["table"],
        "chunk_type": point.payload["chunk_type"],
        "vector_score": point.score,
        "rerank_score": float(score),
        "content": point.payload["content"]
    })
```

```python
reranked = sorted(
    reranked,
    key=lambda x: x["rerank_score"],
    reverse=True
)
```

```python
for item in reranked:
    print(
        f"{item['table']} | "
        f"{item['chunk_type']} | "
        f"vector={item['vector_score']:.3f} | "
        f"rerank={item['rerank_score']:.6f}"
    )
```

```python
# %%
from qdrant_client import QdrantClient
from qdrant_client.models import (
    Distance, VectorParams, SparseVectorParams,
    SparseIndexParams, PointStruct, SparseVector,
    NamedVector, NamedSparseVector
)
from openai import OpenAI
from collections import defaultdict
import uuid

client = QdrantClient(url="http://localhost:6333")

BASE_URL = "https://api.gapgpt.app/v1"
API_KEY  = "sk-s8KnoW59PPxeHBvyzENeVoEiH2QbiNm1PxJt20H586up5p8n"

client_openai = OpenAI(base_url=BASE_URL, api_key=API_KEY)

# %%
# Collection با هر دو نوع vector
client.recreate_collection(
    collection_name="schema_chunk_docs_hybrid",
    vectors_config={
        "dense": VectorParams(
            size=3072,
            distance=Distance.COSINE
        )
    },
    sparse_vectors_config={
        "sparse": SparseVectorParams(
            index=SparseIndexParams(
                on_disk=False
            )
        )
    }
)

# %%

from fastembed import SparseTextEmbedding

sparse_model = SparseTextEmbedding(model_name="Qdrant/bm25")

def get_sparse_vector(text: str) -> SparseVector:
    result = list(sparse_model.embed([text]))[0]
    return SparseVector(
        indices=result.indices.tolist(),
        values=result.values.tolist()
    )

# %%
points = []

for doc in docs:
    # Dense embedding
    dense_emb = client_openai.embeddings.create(
        model="text-embedding-3-large",
        input=doc["text"]
    ).data[0].embedding

    # Sparse embedding
    sparse_emb = get_sparse_vector(doc["text"])

    points.append(
        PointStruct(
            id=str(uuid.uuid4()),
            vector={
                "dense": dense_emb,
                "sparse": sparse_emb
            },
            payload={
                "table": doc["table"],
                "chunk_type": doc["chunk_type"],
                "content": doc["text"]
            }
        )
    )

client.upsert(
    collection_name="schema_chunk_docs_hybrid",
    points=points
)
print(f"Upserted {len(points)} points")

# %%
# Hybrid Search با Reciprocal Rank Fusion (RRF)
from qdrant_client.models import (
    Prefetch, FusionQuery, Fusion,
    SparseVector, NamedVector, NamedSparseVector,
    Query
)

question = "پرفروش ترین محصول رو بده"

# Dense query vector
dense_query = client_openai.embeddings.create(
    model="text-embedding-3-large",
    input=question
).data[0].embedding

# Sparse query vector
sparse_query = get_sparse_vector(question)

# %%
# Hybrid query با RRF fusion
results = client.query_points(
    collection_name="schema_chunk_docs_hybrid",
    prefetch=[
        Prefetch(
            query=dense_query,
            using="dense",
            limit=20
        ),
        Prefetch(
            query=sparse_query,
            using="sparse",
            limit=20
        )
    ],
    query=FusionQuery(fusion=Fusion.RRF),
    limit=15
)

# %%
# نمایش نتایج
for p in results.points:
    print(f"{p.payload['table']} | chunk: {p.payload['chunk_type']} | score: {round(p.score, 4)}")

# %%
# Aggregate بر اساس table
table_scores = defaultdict(float)

for p in results.points:
    table = p.payload["table"]
    table_scores[table] = max(table_scores[table], p.score)

print("\n--- Top Tables ---")
for table, score in sorted(table_scores.items(), key=lambda x: x[1], reverse=True):
    print(f"{table}: {round(score, 4)}")
```
